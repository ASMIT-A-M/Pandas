```python
import pandas as pd

df = pd.read_csv("netflix_titles.csv")


#clean date added column #strip is use to remove spaces from both left and right side
df['date_added'] = df['date_added'].str.strip()


#convert to date format
#september 9 2021 convert to 2021-09-09
df['date_added'] = pd.to_datetime(df['date_added'],errors='coerce')

#Extract year
df['year_added'] = df['date_added'].dt.year

df.head()
```

```python
df.info()
```

```python
df['title'].str.upper()
```

```python
df[df['title'].str.contains('Love',na=False)]
```

```python
df['year_added']=df['year_added'].astype('Int64')
```

```python
print(df['year_added'])
```

```python
df['is_recent'] = df['release_year'].apply(lambda x:"Yes" if x > 2018 else "No")
```
